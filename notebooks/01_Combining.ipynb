# Combining the Datasets

```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path('../data')
OUTPUT_PATH = DATA_DIR / 'HKHJC_FINAL.csv'
```

```python
years = list(range(2009, 2026))
dfs = {}
for year in years:
    path = DATA_DIR / f'HKHJC_{year}.csv'
    dfs[year] = pd.read_csv(path)
    print(f"{year}: {dfs[year].shape}")
```

```python
# Sanity check: every yearly file should share the same column set
# (modulo the stray 'Unnamed: 0' index column dropped below).
columns_per_year = {y: set(d.columns) for y, d in dfs.items()}
core = set.intersection(*columns_per_year.values())
extras = {y: cols - core for y, cols in columns_per_year.items() if cols - core}
print(f"{len(core)} shared columns, extras per year: {extras}")
```

```python
df_big = pd.concat(dfs.values(), ignore_index=True)
df_big = df_big.drop(columns=['Unnamed: 0'], errors='ignore')
print(df_big.shape)
```

```python
df_big.columns.tolist()
```

```python
df_big.to_csv(OUTPUT_PATH, index=False)
print(f"Wrote {OUTPUT_PATH} ({OUTPUT_PATH.stat().st_size / 1e6:.1f} MB)")
```
