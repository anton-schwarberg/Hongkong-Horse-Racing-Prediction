# Combining the Datasets

```python
import pandas as pd
```

```python
years = list(range(2009, 2026))
dfs = {}
```

```python
for element in years:
    filename = f"HKHJC_{element}.csv"
    dfs[element] = pd.read_csv(filename)
```

```python
for df in dfs:
    print(f"{df}: {dfs[df].shape}")
```

```python
df_big = pd.concat(dfs.values(), ignore_index=True)
```

```python
df_big.shape
```

```python
df_big.columns
```

```python
df_new = df_big.drop(columns=['Unnamed: 0'])
```

```python
df_new.to_csv("HKHJC_FINAL.csv", index=False)
```
